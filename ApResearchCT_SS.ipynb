```r
logit_model <- glm(call~education+years_exp, lakisha, family=binomial)
predicted_probs <- predict(logit_model, type = "response")
lakisha <- lakisha%>%mutate(probs=predicted_probs)

scatter_plot(lakisha,
             x= years_exp,
             y= call,
             y_name="Probability of Receiving Callback",
             color_by=education,
             title="Probability of Receiving Callback by Education and Years of Experience",
             subtitle="Colored by Education",
             line=TRUE
             )
```

## **Results**


Now that we've got some visualizations for our data, it is time to use statistics to make empirical conclusions in regards to our research question.

### Logistic Regression: What are the *odds*?


Logistic regression is a tool we use when dealing with variables that are binary, or have only two values. For our purposes, we will be looking at the *call* variable, which takes the values 1 (applicant received a call back) and 0 (applicant did not receive a call back). Ultimately, we would like to learn if an applicant is on average more likely to receive a call back if they are thought to be white, rather than black. ​
​​
​The following code cell has already been written and evaluated for you.​

```r
logistic_regression(lakisha, x=race, y=call)
```

This result tells us that you are roughly **e^(0.4382), or 1.5, times more likely** to receive a call back if you are white applicant, than if your were a black applicant, based on the data of the study.​

Here's something to think about: *Employers want to hire people with X quality, and perhaps the white applicants just have more of those qualities than the black applicants.* **Is this true?** We can again use logistic regression find out. By adding a *covariate*, or another variable into our model, we can effectively control for it. In other words, we can ask the question: assuming we have two applicants with the same X quality (education level, for instance), how much more likely are they to get a call back **soley** due to their race. Use the code below, and any covariate of your choosing, to come to a conclusion:​

```r
logistic_regression(lakisha, x=race, y=call, covariate=years_exp)
```
